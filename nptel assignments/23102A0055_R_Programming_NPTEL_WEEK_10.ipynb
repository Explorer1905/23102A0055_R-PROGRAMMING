## 1. Strings — Substitution (Lecture 40)
`sub(old, new, string)` replaces the **first** match; `gsub(old, new, string)` replaces **all** matches.

```r
y <- "Number of participants: 25"
sub("25", "30", y)
```

```r
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
y
sub("Mr. Singh", "Professor Jha", y)     # only first instance
gsub("Mr. Singh", "Professor Jha", y)    # all instances
```

## 2. Strings — Searching with `grep()` and `grepl()`
`grep()` returns the indices (or values) of matching elements; `grepl()` returns TRUE/FALSE.

```r
str <- c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = TRUE)    # matching elements
grep("ex", str, value = FALSE)   # indices of matches (default)
```

```r
str <- c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str, ignore.case = FALSE, value = TRUE)
grep("R", str, ignore.case = TRUE,  value = TRUE)
grep("R", str, ignore.case = TRUE,  value = FALSE)
grep("R", str, ignore.case = FALSE, value = FALSE)
```

```r
x <- "R course 24.07.2021"
y <- "Number of participants: 25"
c(x, y)
grep("our", c(x, y))    # "our" is in "course" -> element 1
grep("Num", c(x, y))    # "Num" is in "Number" -> element 2
```

```r
str <- c("R Course", "exercises", "include examples of R language")
grepl("R", str)
grepl("ex", str)
```

## 4. Data Frames — More Operations (Lecture 43)

```r
# Summary of a categorical variable = frequency table
summary(painters$School)
```

### `attach()` and `detach()`

```r
attach(painters)
summary(School)         # categorical variable
summary(Composition)    # numeric variable
detach(painters)
```

```r
# After detach() the variable must be referred as painters$School again
res <- try(summary(School), silent = TRUE)
cat(res)
summary(painters$School)
```

### Subsets — `subset()` and logical indexing

```r
subset(painters, School == "F")
```

```r
# Equivalent command
painters[painters[["School"]] == "F", ]
```

```r
subset(painters, Composition <= 6)
```

```r
# Eliminate uninteresting columns (3rd = Colour, 5th = School)
subset(painters, School == "F", select = c(-3, -5))
```

### `split()` — partition a data frame by a factor

```r
splitted <- split(painters, painters$School)
names(splitted)
splitted$A
```

```r
splitted$F
is.data.frame(splitted$A)
```

## 5. Data Frames — Combining and Merging (Lecture 44)
- `cbind()` — combine columns side by side
- `merge()` — join two data frames using a common column
- `rbind()` — stack data frames on top of each other

```r
df1 <- data.frame(state = c("UP", "MP", "AP", "JK"),
                  popnsize = c(1000, 2000, 3000, 4000))

df2 <- data.frame(state = c("UP", "MP", "AP", "JK"),
                  samplesize = c(100, 200, 300, 400),
                  surveycompleted = c("Yes", "No", "Yes", "No"))
df1
df2
```

### `cbind()`

```r
cbind(df1, df2)     # note: 'state' column appears twice
```

### `merge()`

```r
merge(df1, df2, by = "state")   # rows sorted by the key
```

### `rbind()`

```r
df11 <- data.frame(state = c("UP", "MP", "AP", "JK"),
                   popnsize = c(1000, 2000, 3000, 4000))
df22 <- data.frame(state = c("Bihar", "Delhi", "Punjab"),
                   popnsize = c(100, 200, 300))
df11
df22
rbind(df11, df22)
```
