### Lecture 36: Print and Format with Concatenation

### `print()` limitation trying to print multiple items gives an error

```r
print("The zero occurs at", 2*pi, "radians.")
```

### The only way with `print` is to print items one at a time

```r
print("The zero occurs at"); print(2*pi); print("radians")
```

### `cat()` combine multiple items into continuous output

```r
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

### `cat()` with `date()`

```r
d = date()
cat("Today\u2019s date is:", d, "\n")
```

### `cat()` with `sep` argument on a vector

```r
x = 1:10
x
cat(x, sep =" ++ ")
cat("\n")   # This is used to change line
cat(x, sep =" / ")
```

### `cat()` combining text and computed values

```r
x = 7
cat("The square of", x, "is", x^2, "!\n")
```

### `cat()` with `format()`

```r
cat("The square root of",x,"is approximately", format(sqrt(x),digits=3),"\n")
```

### `cat()` can also print simple vectors

```r
evenno = c(2,4,6,8,10)
evenno
cat("The first few even numbers are:", evenno, "...\n")
```

### `cat()` with `fill` and `labels`

```r
x = 1:10
x
cat(x, fill = 2, labels = paste("(", letters[1:10], ")"))
```

## Lecture 37: Paste Function

### `paste()` converts arguments to character strings (via `as.character`)

```r
paste(1:12)
```

```r
as.character(1:12)   #Alternative to paste
```

### `paste()` with default separator (space)

```r
paste("Everybody", "loves", "R Programming.")
```

### `paste()` with custom `sep`

```r
paste("Everybody", "loves", "R Programming.", sep="*")
```

```r
paste("Everybody", "loves", "R Programming.", sep="===")
```

### `paste()` with vectors of strings generates all combinations

```r
names = c("Prof. Singh", "Mr.  Venkat", "Dr. Jha")
names
```

```r
paste(names, "is", "a good", "person.")
```

### `paste()` with `collapse` join all combinations into one string

```r
paste(names, "is", "a good", "person.", collapse=", and ")
```

### Role of blank space inside double quotes with `sep`

```r
paste(1, " is first", 2, " is second", 3, " is third", sep = "#")
```

```r
paste(1, " is first   ", 2, " is second   ", 3, " is third   ", sep = "#")
```

### `paste()` producing a vector of strings; indexing elements

```r
x = paste("Ex", 1:5, sep="_")
x
x[1]
x[2]
x[3]
x[5]
```

### `paste()` with `collapse` producing a single string instead of a vector

```r
x = paste("Ex", 1:5, sep="_", collapse="")
x[1]
```

### Difference between `paste` with and without `collapse`

```r
x = paste("Ex", 1:5, sep="_")
x
```

```r
x = paste("Ex", 1:5, sep="_", collapse="")
x
```

### `paste0()` vs `paste()` for a single vector

```r
paste0(1:10)
```

```r
paste(1:10)
```

### `paste0()` and `paste()` with more than one vector (vectorized concatenation)

```r
paste0(1:10, c("st", "nd", "rd", rep("th", 7)))
```

```r
paste(1:10, c("st", "nd", "rd", rep("th", 7)))
```

## Lecture 38: String Splitting

### `strsplit()` basics

```r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"
x
```

```r
strsplit(x, split="!")
```

### `strsplit()` with a multi-character split string

```r
strsplit(x, split = "&!")
```

### `strsplit()` result stored in a variable and accessing components

```r
y = strsplit(x, split = "!&")
y
```

```r
y[[1]][1]
```

```r
y[[1]][2]
```

```r
y[[1]][3]
```

### Example: Convert split dates to a matrix

```r
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
dates
```

```r
datesplt = strsplit(dates, "-")
datesplt
```

```r
datemat = matrix(unlist(datesplt), nrow = 4, ncol=3, byrow=TRUE)
datemat
```

```r
datematrix = matrix(as.numeric(unlist(datesplt)), nrow = 4, ncol=3, byrow=TRUE)
datematrix
```

### `strsplit()` splitting into individual characters

```r
strsplit("Shalabh",split="")
```

## Lecture 39: Manipulations with Strings and Alphabets

### `nchar()` — count characters

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"
nchar(x)  #Count the Number of Characters
nchar(y)  #Count the Number of Characters
```

### `nzchar()` — check for non-empty strings

```r
nzchar(x)
nzchar(y)
```

### `nzchar()` on a character vector

```r
x = c("Apple", "Banana", "Cake")
x
nzchar(x)
```

```r
y = c("Apple", "", "Cake")
y
nzchar(y)
```

### `nchar()` on character and numeric vectors

```r
x = c("Apple", "Banana", "Cake")
nchar(x)
```

```r
y = c(2, 4, 6)
nchar(y)
```

```r
z = c(11, 222, 3333)
nchar(z)
```

```r
z1 = c(1.1, 2.22, 3.333)
nchar(z1)
```
