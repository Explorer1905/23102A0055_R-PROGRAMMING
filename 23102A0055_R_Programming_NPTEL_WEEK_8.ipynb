## 1. Operations on Lists (Lecture 31)

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")
list1
list2
```

### Merging lists

```r
list12 <- c(list1, list2)
list12
```

### Converting a list to a vector — `unlist()`

```r
unlist(list1)
unlist(list2)
mode(list1)
mode(unlist(list1))
```

### Appending to a list — `append()`

```r
append(list1, 100)
append(list2, "coffee")
```

```r
# Append after a given position
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

### Removing from a list — `listname[-position]`

```r
list1[-2]
list2[-1]
```

### Extracting from a list — using a range of indexes

```r
list1 <- list(1, 2, 3, 4, 5, 6)
list2 <- list("water", "juice", "lemonade", "tea", "coffee", "milk")

list1[2:4]
list1[c(1, 3, 5)]
list2[2:4]
list2[c(1, 3, 5)]
```

## 2. Vector Indexing (Lecture 32)

```r
x <- 1:10
x
x[x > 5]
x[x %% 2 == 0]   # %% is the mod operator: even numbers
x[x %% 2 == 1]   # odd numbers
```

### Vector with missing observations

```r
x[5] <- NA
x
y <- x[!is.na(x)]   # ! means negation
y
mean(x)
mean(y)
```

### Negative integers and the empty index

```r
x <- 1:10
x[-(1:5)]
x[6:10]     # same outcome
x[]         # empty index returns the whole vector
```

### Named elements

```r
z <- list(a1 = 1, a2 = "c", a3 = 1:3)
z
names(z)

# Change only the name of the third element
names(z)[3] <- "c2"
z
```

```r
x <- c(water = 1, juice = 2, lemonade = 3)
names(x)
x["juice"]
```

### Mixed mode — a list with dimensions

```r
ab <- list(1, 2, 3, "X", "Y", "Z")
dim(ab) <- c(2, 3)
print(ab)
mode(ab)
```

## 3. Factors (Lecture 33)
A **factor** represents a categorical variable. Categories (levels) are stored internally as integer codes with labels giving meaningful names.

```r
# Ball colours: Red = 1, Blue = 2, Green = 3
balls <- c("Red", "Green", "Green", "Blue", "Red")
fb <- factor(balls, levels = c("Red", "Blue", "Green"))
fb
levels(fb)
as.integer(fb)
```

### Die-roll example using `levels` and `labels`

```r
y <- c(1, 4, 3, 5, 4, 2, 4)                      # seven rolls
possible.dieface <- c(1, 2, 3, 4, 5, 6)
labels.dieface   <- c("one", "two", "three", "four", "five", "six")

facy <- factor(y, levels = possible.dieface, labels = labels.dieface)
facy
```

```r
# Difference between a plain vector and a factor
y
facy
table(facy)
```

## 4. Factors — `as.factor`, `class` and `unclass` (Lecture 34)

```r
x <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
y <- as.factor(x)
y            # levels ordered numerically
```

```r
x <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
x            # levels ordered alphabetically
```

### `class()`

```r
class(9)
class("9")
class(print)
class(matrix(nrow = 2, ncol = 2, data = 1:4))
```

### `unclass()` — removes the class and shows the integer codes

```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)
brands_fac
unclass(brands_fac)
```

```r
# Use the codes to index another vector
colours <- c("blue", "green", "red")
colours[unclass(brands_fac)]
```

```r
x <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
unclass(x)

# Different assignment of codes using levels
x <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"),
            levels = c("water", "juice", "lemonade"))
x
unclass(x)
levels(x)
```

### Ordered factor

```r
income <- ordered(c("high", "high", "low", "medium", "medium"),
                  levels = c("low", "medium", "high"))
income
unclass(income)
income[1] > income[3]
```

## 5. Strings — Display and Formatting (Lecture 35)
Important commands: `print`, `format`, `cat`, `paste`.

### `print()`

```r
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)
```

```r
print("apple")
print(c("apple", "banana"))
print(c("apple", "banana", 6, 10))   # numbers are converted to character
```

### `format()` — digits, nsmall, width and justify

```r
print(format(0.5, digits = 10, nsmall = 15))
```

```r
s <- c("A", "BB", "CCC", "DDDD")
format(s, width = 7, justify = "centre")
format(s, width = 7, justify = "left")
format(s, width = 7, justify = "right")
format(s, width = 7, justify = "none")
```

### Printing a matrix

```r
x <- matrix(nrow = 3, ncol = 2, data = 1:6, byrow = TRUE)
print(x)
```

### Large quantities — `big.mark`

```r
format(1234567, big.mark = ",")
format(12345678, big.mark = ",")
format(123456789, big.mark = ",")
format(123456789, big.mark = " ")
```

### `cat()` and `paste()`

```r
cat("The square root of 2 is", round(sqrt(2), 4), "\n")
paste("Item", 1:3, sep = "-")
```
