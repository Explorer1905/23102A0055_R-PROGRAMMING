## 1. What is R?
R is an environment for data manipulation, statistical computing, graphics display and data analysis. It is free, open source, interpreted (each command is evaluated as soon as it is entered) and available for Windows, Linux and macOS.

## 2. Getting Help in R
There are several ways to get help: `?function`, `help()`, `help.search()`, `help.start()`, `find()`, `apropos()`, `example()` and `demo()`.

```r
# 1) Help on a particular function (same as help("read.table"))
?read.table
```

```r
# 2) Search help pages by subject
help.search("data input")
```

```r
# 3) help() for a function
help(mean)
```

```r
# 4) find(): which package contains a function?
find("lowess")
```

```r
# 5) apropos(): names of all objects matching a pattern
apropos("lm")
```

```r
# 6) Worked example of a function
example(mean)
```

```r
# 7) demo(): see what R can do (uncomment to try, it is interactive)
# demo(graphics)
# demo(persp)
```

## 3. Packages and Libraries
- A **package** is a collection of functions and data sets bundled together.
- A **library** is the place where packages are stored; `library()` loads a package.
- Some packages come with base R (e.g. `MASS`, `mgcv`, `spatial`), others (e.g. `boot`, `cluster`) must be installed as per need.

```r
# Packages that come with the standard R installation
library(MASS)
library(spatial)

# List of attached packages
search()
```

```r
# Description of a package
packageDescription("spatial")
```

```r
# Contents of a package
library(help = "spatial")
```

### Installing packages
`install.packages()` downloads and installs a package from CRAN. The code below installs a package only if it is not already available.

```r
if (!requireNamespace("boot", quietly = TRUE)) install.packages("boot")
if (!requireNamespace("cluster", quietly = TRUE)) install.packages("cluster")

library(boot)
library(cluster)
```

```r
# Packages installed on this computer
ip <- installed.packages()
head(ip[, c("Package", "Version")], 10)
nrow(ip)
```

## 4. Command Line versus Scripts
- R commands are typed (not menu driven); single-line and multi-line commands are possible.
- For multi-line programs it is better to use a script/text editor such as the RGui editor or **RStudio** (free IDE for R, https://www.rstudio.com).
- A script is a saved file of commands (`.R`) that can be run again.

```r
# A multi-line command
total <- 0
for (i in 1:5) {
  total <- total + i
}
total
```

```r
# Save a few commands in a script file and run it with source()
writeLines(c("a <- 5", "b <- 7", "cat('Sum =', a + b, '\\n')"), "my_script.R")
source("my_script.R")
```
